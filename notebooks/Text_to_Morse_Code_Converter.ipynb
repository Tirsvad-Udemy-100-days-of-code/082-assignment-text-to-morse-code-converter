## Step 2: The Morse table

A dictionary holds the signal pattern (dots and dashes) of each letter, digit and common punctuation mark. It lives in `constants.py` of the Domain.

```python
from morse_converter import MORSE_CODE_DICT

print(len(MORSE_CODE_DICT), "characters")
for char in "ESOT019?":
    print(char, MORSE_CODE_DICT[char])
```

## Step 3: Convert text to Morse code

`text_to_morse` splits the text into words, joins the signal patterns of a word with one space and the words with ` / `. Upper and lower case give the same result, and a blank text gives an empty Morse code.

```python
from morse_converter import text_to_morse

print(text_to_morse("SOS"))
print(text_to_morse("Hello World"))
print(repr(text_to_morse("")))
```

A character with no signal pattern rejects the whole text (decision [ADR-0001](../docs/adr/adr-0001-unsupported-characters-are-rejected.md)):

```python
from morse_converter.domain.errors import UnsupportedCharacterError

try:
    text_to_morse("good ~ text")
except UnsupportedCharacterError as error:
    print("Not supported:", repr(error.char))
```

## Step 4: Run it from the command line

`python -m morse_converter` takes the text from the arguments or asks for it at a prompt. It prints the Morse code and exits with 0, or names the unsupported character on the error stream and exits with 1.

```python
import subprocess

done = subprocess.run(
    [sys.executable, "-m", "morse_converter", "Hello", "World"],
    capture_output=True,
    text=True,
    check=False,
)
print(done.stdout.strip(), "| exit code", done.returncode)
```
